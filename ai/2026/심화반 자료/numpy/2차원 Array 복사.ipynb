```python
## 리스트의 경우
a = [1, 2, 3, 4]
b = a
id(a), id(b), a is b
```

```python
b[0] = 999 # a와 b 는 여전히 같은 객체
id(a), id(b), a is b
```

```python
## 다른 객체로 복사하기
a = [1, 2, 3, 4]
b = a.copy()
a is b, id(a), id(b) # 다른 객체로 복사
```

```python
b[0] = 999
b
```

```python
a
```

```python
a[0] = 1
a
```

```python
## 2차원 리스트의 경우
a = [[1, 2, 3],
    [4, 5, 6]]
b = a.copy()
id(a), id(b), a is b # 다른 객체로 생성
```

```python
b[0][0] = 999
b
```

```python
a
```

```python
## 2차원 이상의 리스트에서 다른 객체로 복사하기
import copy
b = copy.deepcopy(a)
b
```

```python
id(a), id(b), a is b # 다른객체
```

```python
b[0][0] = 999
b
```

```python
a # a 내용은 바뀌지 않고 그대로 유지
```

```python
id(a[0][0]), id(b[0][0]), a[0][0] is b[0][0]
```
