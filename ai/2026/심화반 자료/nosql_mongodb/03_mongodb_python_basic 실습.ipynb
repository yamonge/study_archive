### 4. 데이터 삽입
- 데이터를 MongoDB에 삽입하려면 insert_one() 또는 insert_many() 메서드를 사용합니다.

````python
db = client['test2'] # use test2 (데이터 베이스 선택 or 생성)
db
````

````python
collection = db.users2 # db.createCollection("users2") 컬렉션 새로 생성
collection
````

````python
data = {"name": "John", "age": 30}
result = collection.insert_one(data)   # db.users.insertOne( { name: "John", age: 30 } )
print("Inserted ID:", result.inserted_id) # _id 
````

````python
# 여러 문서 삽입
data = [
    { "name": "Alice", "age": 25 }, 
    { "name": "Bob", "age": 35 }, 
    { "name": "Charlie", "age": 40 }
]
result = collection.insert_many(data)  # db.users.insertMany( [ { name: "Alice", age: 25 }, 
                                       #                        { name: "Bob", age: 35 }, 
                                       #                        { name: "Charlie", age: 40 } ] )
result
````

### 5. 데이터 조회
- 데이터를 조회하려면 find_one() 또는 find() 메서드를 사용합니다.

> MongoDB 에서는 findOne() 또는 insertMany 과 같이 naming 이 되어 있지만, pymongo 에서는 find_one() 또는 insert_many() 와 같은 naming 으로 되어 있음

````python
# 단일 문서 조회
db = client['test2'] # db 선택
collection = db.user2 # 컬렉션 선택 
document = collection.find_one({"name": "John"})
document
````

````python
# 모든 문서 조회 (find())
documents = collection.find() # db.users2.find()
documents
````

````python
list(documents) # find() 결과는 cursor 객체이므로 list()로 변환하여 출력
````

````python
documents = collection.find() # db.users2.find()
for doc in documents:
    print(doc)
````

### 6. 데이터 수정
- 데이터를 수정하려면 update_one() 또는 update_many() 메서드를 사용합니다.

<img src="https://davelee-fun.github.io/fixeddata/mongodb_update_structure.png" /> 

````python
# 단일 문서 수정(update_one)
db = client['test2'] # db 선택
collection = db.user2 # 컬렉션 선택 
filter = {"name": "John"} # 수정할 문서 선택 (name이 "John"인 문서)
update = {"$set" : {"age" : 31}} # 수정할 내용
result = collection.update_one(filter, update) # db.users2.updateOne({name:"John"}, {$set: {age:31}})
result.modified_count # 수정된 문서 수
````

````python
db
````

````python
# 여러 문서 수정
filter = {"age": {"$gt": 30}}
update = {"$set": {"is_available": True}}
result = collection.update_many(filter, update) # db.users2.updateMany({age: {$gt: 30}}, {$set: {is_available}})
result.modified_count
````

````python
documents = collection.find()
documents
````

### 7. 데이터 삭제
- 데이터를 수정하려면 delete_one() 또는 delete_many() 메서드를 사용합니다.

````python
from pymongo import MongoClient
client = MongoClient("mongodb://localhost:27017/")
db = client['test2']
collection = db.user2 
docs = collection.find()
for doc in docs:
    print(doc)
````

````python
filter = {"name":"John"}
result = collection.delete_one(filter) # db.user2.deleteOne({name: "John"})
result.deleted_count # 삭제된 문서 수
````

````python
docs = collection.find()
for doc in docs:
    print(doc)
````

### 8. 프로그램 종료
- 프로그램 종료시에는 MongoClient() 객체에 close() 를 명시적으로 호출해주는 것이 좋습니다.
- 또는 다음과 같이 파이썬의 with 문법을 활용하면, close() 를 명시적으로 호출해주지 않아도, 자동으로 호출됨

```python
    from pymongo import MongoClient

    # MongoDB에 연결
    with MongoClient("mongodb://localhost:27017") as client:
        # 작업 수행
        pass
```

````python
# 연결 종료
# client = MongoClient("mongodb://localhost:27017/") # MongoDB 연결
client.close()
````

### 9. 전체 pymongo 템플릿 코드

````python
from pymongo import MongoClient

# MongoDB에 연결 (인증 미필요시)
client = MongoClient("mongodb://localhost:27017")
# client = MongoClient("mongodb://username:password@localhost:27017")
# 인증이 필요하지 않은 경우 위의 첫 번째 줄 사용, 인증이 필요한 경우 두 번째 줄 사용

db = client['test2']  # 'test' 데이터베이스 선택

collection = db.user2  # 'users' 컬렉션 선택

documents = collection.find()  # 'users' 컬렉션의 모든 문서 조회
for document in documents: # find() 의 결과는 iterable 객체이므로 반복문을 통해 각 데이터를 가져와야 함
    print(document)

# 연결 종료
client.close()
````

````python
from pymongo import MongoClient

# with 구문을 사용하여 MongoDB 연결 (자동으로 연결 종료)
with MongoClient("mongodb://localhost:27017") as client:
    db = client['test2']  # 'test2' 데이터베이스 선택
    collection = db.user2  # 'users2' 컬렉션 선택
    documents = collection.find()  # 'users2' 컬렉션의 모든 문서 조회
    for document in documents: # find() 의 결과는 iterable 객체이므로 반복문을 통해 각 데이터를 가져와야 함
        print(document)
````
