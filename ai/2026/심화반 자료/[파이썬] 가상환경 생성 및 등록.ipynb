## 2. Step 1: Conda 가상환경 생성하기

프로젝트 전용 '독립된 공간(Environment)'을 만듭니다.

### 🔹 명령어 구조
```bash
    conda create -n [환경이름] python=[버전]
```
### 🔹 실습 예제
터미널에서 아래 명령어를 입력하세요. (`Proceed [y]/n?` 질문이 나오면 **y** 입력)

```bash
    conda create -n my_lab python=3.10
```

## 3. Step 2: 가상환경 활성화 및 패키지 설치

방을 만들었으면 그 방으로 **들어가야(Activate)** 합니다. 이후 주피터 사용을 위한 도구를 설치합니다.

### 🔹 가상환경 진입
터미널(Anaconda Prompt)에서 아래 명령어를 입력합니다.

```bash
    conda activate my_lab
```

* **Check:** 프롬프트 앞부분이 `(base)` → `(my_lab)`으로 바뀌었는지 반드시 확인하세요.

### 🔹 필수 패키지 설치
`ipykernel`은 VS Code와 가상환경을 연결해주는 **통역사 역할**을 합니다. 반드시 설치해야 합니다.

```bash
    pip install jupyter ipykernel
```

---



## 4. Step 3: Jupyter Kernel 등록

가상환경을 만들었어도 등록하지 않으면 VS Code 목록에 뜨지 않습니다. 시스템에 **이름표**를 달아주는 과정입니다.

### 🔹 명령어 구조
```bash
    python -m ipykernel install --user --name [내부이름] --display-name "[표시이름]"
```
* `--name`: 시스템 내부 식별용 (가상환경 이름 추천)
* `--display-name`: VS Code 화면에 보일 이름

### 🔹 실습 예제
```bash
    python -m ipykernel install --user --name my_lab --display-name "Python 3.10 (My Lab)"
```

* **성공 메시지:** `Installed kernelspec my_lab in ...` 이 뜨면 성공입니다.

## 5. Step 4: VS Code에서 연동 및 테스트

이제 VS Code를 열고 방금 등록한 커널을 선택해 봅니다.

### 🔹 1. VS Code 실행 및 파일 생성
1. **VS Code**를 실행합니다.
2. 빈 폴더를 열고 새로운 파일 `test.ipynb`를 생성합니다.

### 🔹 2. 커널 선택 (Select Kernel)
1. 우측 상단의 **[Kernel 선택]** 또는 **[Select Kernel]** 버튼을 클릭합니다.
2. **Jupyter Kernel...** 을 선택합니다.
3. 목록에서 방금 등록한 **`Python 3.10 (My Lab)`** 을 찾아 클릭합니다.

### 🔹 3. 동작 확인
아래 코드를 셀에 입력하고 `Shift + Enter`로 실행합니다.

```python
    import sys
    print(sys.executable)
```

> **✅ 성공 기준:** 출력된 경로에 `.../envs/my_lab/...` 처럼 내가 만든 가상환경 폴더명이 포함되어 있다면 연동 성공입니다!
