```python
import pandas as pd
import numpy as np
```

```python
obj = pd.Series([4,7,-5,3])
obj
```

```python
obj.index
```

```python
obj.shape
```

```python
obj2 = pd.Series([4,7,-5,3], index=['d','b','c','a'])
obj2
```

```python
# Series는 dictionary 와 같이 사용 가능
sdata = {
    'charles': 35000,
    'hyunsu': 35000,
    'hayoung': 35000,
    'joohee': 35000
}
```

```python
obj3 = pd.Series(sdata)
obj3
```

```python
obj3.name = 'Salary'
obj3.index.name = 'name'
obj3
```

```python
obj3.index=['A','B','C','D']
obj3
```

# 2) DataFrame
pd.DataFrame() 함수를 사용해서 정의. 파이썬의 딕셔너리, numpy array, list등 데이터 프레임으로 활용 가능

```python
data = {"names" : ["aa","aa","aa","bb","bb"],
        "years" : [2014,2015,2016,2017,2018],
        "points" : [1.5,1.7,3.6,2.4,2.9]}

df = pd.DataFrame(data)
df
```

```python
df.index
```

```python
df.columns
```

```python
df.values
```

```python
df.describe()       # 계산이 가능한 컬럼에 한해서 각 컬럼의 평균,분산,최소/최대값 등 기본 통계량을
                    # 산출한 결과를 보여줌... 데이터를 읽어온 후, 데이터 셋을 전체적으로 살펴보고자
                    # 할때 사용하기 유용함
```

```python
df2 = pd.DataFrame(data, columns=["years","names","points","penalty"],
                   index = ["one","two","three","four","five"],
                   )
df2
```

```python
df2.index
```

```python
df2.columns
```

```python
df2.values
```

```python
df2.describe()
```
