```python
arr = np.random.randn(8)
arr
```

```python
# 1차원 오름차순 정렬
np.sort(arr)
```

```python
# 1차원 내림차순 정렬
np.sort(arr)[::-1]
```

```python
# 2차원 오름차순 정렬
arr = np.random.randn(5,3)
arr
```

```python
# 행 방향으로 열의 값을 오름차순 정렬
np.sort(arr, axis=0)
```

```python
# 열방향으로 행의 값을 오름차순 정렬
np.sort(arr, axis=1)
```

```python
# Large_arr 어레이가 정의가 되어 있을 때, 해당 array 상에서 상위 5%에 위치하는 값을 찾기
large_arr = np.random.randn(150)
large_arr
```

```python
# 내림차순 정렬
np.sort(large_arr)[::-1]
```

```python
# 상위 5% 값 찾기(0.05 * 150)
np.sort(large_arr)[::-1][:int(0.05 * len(large_arr))] 
```
