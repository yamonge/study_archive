```python
# Subject.scv 파일을 읽어 와서 'df' 변수에 저장하기
data = pd.read_csv("../12/Subjects.csv", encoding="utf-8")
df2 = pd.DataFrame(data)
df2
```

```python
# df DataFrame 에 '총점' 과 '평균' 열을 추가하여 각 학생 별 총점, 평균 점수를 입력
df2.loc[:, 'total']  = None
df2.loc[:, 'avg'] = None
df2
# 행 삭제
# df.drop(index=['total', 'avg'], inplace=True)
# df

```

```python
for i in df2.index:
  total = np.sum(df2.loc[i, ['kor', 'eng', 'math', 'science']])
  avg = round(total / 4, 2)
  df2.loc[df2.index[i], ['total', 'avg']] = [total, avg]

df2
```

```python
# class 1과 class 2 의 평균 점수를 각각 출력
class1 = df2.loc[df2['class'] == 1, :]
class2 = df2.loc[df2['class'] == 2, :]
class1
class2
```

```python
if np.mean(class1.loc[:, 'avg']) > np.mean(class2.loc[:, 'avg']):
  print(f"1반이 짱 : {np.mean(class1.loc[:, 'avg'])}")
else:
  print(f"2반이 짱 : {np.mean(class2.loc[:, 'avg'])}")
  
```
