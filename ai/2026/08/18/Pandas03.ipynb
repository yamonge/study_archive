```python
# 상관계수 구하기
df2 = pd.DataFrame(np.random.randn(6,4))
df2
```

```python
df2.columns = ['a', 'b', 'c', 'd']
df2.index = pd.date_range('20151201', periods = 6)
df2
```

```python
# a열과 b열의 상관계수 산출하기(피어슨상관계수로 계산)
df2['a'].corr(df2['b'])
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
df2[['a', 'b']].plot(figsize=(10,5))

plt.title("random time series data")
plt.xlabel("data")
plt.ylabel("value")
plt.grid(True)
```

```python
df2.corr()
```

```python
# 정렬
# 불규칙 데이터 섞기
dates = df2.index
random_dates = np.random.permutation(dates)
df2 = df2.reindex(index=random_dates, columns=['d','c','a','b'])
df2
```

```python
df3 = pd.DataFrame(np.random.randn(4,3), 
                   index=['seoul','incheon','busan','daegu'],
                   columns=['b','d','e'])

df3
```

```python
func = lambda x: x.max() - x.min()        # 행 또는 열의 최대값,최소값
```

```python
df3.apply(func, axis=0)     # 열 별로 값을 생산
```

```python
#print(func(df3.loc[:, 'b']))
print(func(df3['b']))
```

```python
df3.apply(func, axis=1)
```
