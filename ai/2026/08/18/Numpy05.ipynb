```python
import numpy as np
```

```python
data = np.genfromtxt('./data/ratings.dat', delimiter='::', dtype=np.int64)
data
```

user_id :: item :: rating :: timestamp

# 문제) 전체 평균 평점 계산하기

```python
avg = np.mean(2)
avg
```

# 문제) 각 사용자 별 평균 평점 계산하기

```python
user_id = data[:, 0]
unique_user_id = np.unique(user_id)
mean_rating_by_user = []
for user_id in unique_user_id:
  user_ratings = data[data[:, 0] == user_id, 2]
  rating_mean = np.mean(user_ratings)

  mean_rating_by_user.append([user_id, rating_mean])

mean_rating_by_user = np.array(mean_rating_by_user)
mean_rating_by_user
```

# 결과를 파일에 writing하여 저장하기

```python
np.savetxt('./data/mean_rating_by_user.csv', mean_rating_by_user, fmt=['%d','%.3f'], delimiter=',')
```

# 상위 5% 결과 띄우기

```python
sort_user = np.argsort(mean_rating_by_user[:, 1])[::-1]
sort_users = mean_rating_by_user[sort_user][:int(len(mean_rating_by_user) * 0.05)]
```

```python
np.savetxt('./data/mean_rating_by_user.csv', sort_users, fmt=['%d','%.3f'], delimiter=',')
```
